```python
import numpy as np
```

```python
# Load your dataset
#df = pd.read_csv("your_plants_data.csv")

# 1. Clean scientific names
def clean_scientific_name(row):
    if pd.isna(row['nombreCientifico_snib']) or 'nan' in str(row['nombreCientifico_snib']).lower():
        # Reconstruct from taxonomic components
        parts = [
            row['genero_snib'],
            row['epitetoespecifico_snib'],
            row['epitetoinfraespecifico_snib']
        ]
        return ' '.join([p for p in parts if pd.notna(p)])
    return row['nombreCientifico_snib']

df['cleaned_scientific_name'] = df.apply(clean_scientific_name, axis=1)

## reino_snib  , phylum_snib ,  clase_snib ,orden_snib , familia_snib

# 2. Fill higher taxonomy from scientific name
taxonomic_ranks = ['reino_snib'  , 'phylum_snib' ,  'clase_snib' ,'orden_snib' , 'familia_snib']

for rank in taxonomic_ranks:
    col_name = f'{rank}'
    # Fill missing values with most common rank for that scientific name
    df[col_name] = df.groupby('cleaned_scientific_name')[col_name].transform(
        lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan))

    # Fill remaining with most common for genus
    df['genus'] = df['cleaned_scientific_name'].str.split().str[0]
    df[col_name] = df.groupby('genus')[col_name].transform(
        lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown'))

# 3. Create compact taxonomy path
df['taxonomic_path'] = df[taxonomic_ranks].apply(lambda x: ' | '.join([str(i) for i in x if pd.notna(i)]), axis=1)

# Show cleaned data
print(f"Cleaned scientific names: {df['cleaned_scientific_name'].nunique()} unique species")
print(df[['cleaned_scientific_name', 'taxonomic_path']].head())
```

Taxonomic Clustering (Natural Groups)

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import AgglomerativeClustering

# Encode taxonomic paths
encoder = LabelEncoder()
taxonomic_encoded = encoder.fit_transform(df['taxonomic_path'])

# Hierarchical clustering
clusterer = AgglomerativeClustering(n_clusters=15, metric='euclidean')
df['taxonomic_cluster'] = clusterer.fit_predict(taxonomic_encoded.reshape(-1, 1))
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt
```

```python
# Encode taxonomic paths
encoder = LabelEncoder()
encoded_taxonomy = encoder.fit_transform(df['taxonomic_path'])

# Calculate linkage matrix
# Using 'ward' method which is suitable for Euclidean distances
Z = linkage(encoded_taxonomy.reshape(-1, 1), method='ward')

# Plot truncated dendrogram
plt.figure(figsize=(15, 10))
dendrogram(Z,
           p=30,  # Display only the last 'p' merged clusters
           truncate_mode='lastp', # Truncate to show only the last 'p' merges
           labels=df['cleaned_scientific_name'].values, # Keep original labels (will be truncated)
           orientation='right',
           leaf_font_size=10, # Adjust font size for truncated view
           color_threshold=150)
plt.title('Taxonomic Clustering Dendrogram (Truncated)')
plt.xlabel('Distance')
plt.ylabel('Plant Species or Cluster Size') # Label update to reflect truncation
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px

# Create sunburst data
sunburst_df = df.groupby(['reino_snib', 'phylum_snib', 'clase_snib',
                          'orden_snib', 'familia_snib', 'taxonomic_cluster']).size().reset_index(name='count')

# Generate visualization
fig = px.sunburst(
    sunburst_df,
    path=['reino_snib', 'phylum_snib', 'clase_snib', 'orden_snib', 'familia_snib', 'taxonomic_cluster'],
    values='count',
    title='Taxonomic Distribution by Cluster',
    hover_data=['count']
)
fig.update_traces(textinfo="label+percent parent")
fig.show()
```

Network


```python
import networkx as nx
import matplotlib.pyplot as plt
from community import community_louvain

# Create graph
G = nx.Graph()

# Add nodes with cluster attributes
for _, row in df.iterrows():
    G.add_node(row['cleaned_scientific_name'],
               cluster=row['taxonomic_cluster'],
               family=row['familia_snib'])

# Add edges between same-family species
for family in df['familia_snib'].unique():
    family_species = df[df['familia_snib'] == family]['cleaned_scientific_name']
    for i, sp1 in enumerate(family_species):
        for sp2 in list(family_species)[i+1:]:
            G.add_edge(sp1, sp2, weight=3)

# Visualize
plt.figure(figsize=(20, 15))
pos = nx.spring_layout(G, seed=42)
colors = [G.nodes[n]['cluster'] for n in G.nodes]

nx.draw_networkx_nodes(G, pos, node_size=50, node_color=colors, cmap=plt.cm.tab20)
nx.draw_networkx_edges(G, pos, alpha=0.1)
plt.title('Taxonomic Cluster Network')
plt.show()
```
